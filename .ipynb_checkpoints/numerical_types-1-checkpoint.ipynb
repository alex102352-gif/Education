This notebook explores basic numerical types in Python and how they behave.

## 1. Basic Types
Python has several built-in numeric types.

```
type(4), type(4.0), type(3+4j)
```

## 2. Integers vs Floats
- Integers: no decimal
- Floats: decimal values

```
x = 4
y = 4.0

print(type(x))
print(type(y))
```

## 3. Large Integers
Python supports very large integers.

```
2**50
2**200
```

## 4. Division Behavior
Python uses true division by default.

```
2 / 3
```

Use floor division for integer results.

```
2 // 3
5 // 3
```

## 5. Modulus (Remainder)

```
5 % 3
```

## 6. Float Representation

```
1.5
1e-6
```

## 7. Type Conversion

```
int(1.6)
float(3)
```

Note: int() truncates, it does NOT round.

## 8. Floating Point Precision Issues

```
0.1 + 0.2 == 0.3
```

Let's inspect actual stored values.

```
format(0.1, '.17f')
format(0.2, '.17f')
format(0.3, '.17f')
```

## 9. Exact Representation Example

```
0.5 + 0.125 == 0.625
```

These work because they can be exactly represented in binary.

## 10. Key Takeaways
- Python integers are unbounded
- Floats are approximate
- Avoid exact equality with floats
- Use // for integer division
- Use % for remainder
