# Python Exceptions


This notebook introduces errors, specific exception handling, `raise`, and the `else` and `finally` clauses.

## 1. A Common Exception
Dividing by zero raises a `ZeroDivisionError`.

```python
# 1 / 0  # Uncommenting this line raises ZeroDivisionError

print("The line above is commented out so the notebook can continue.")
```

## 2. Catch a Specific Exception
Use `try` for code that may fail and `except` for the specific error you want to handle.

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError as error:
        print("Cannot divide by zero:", error)

print(divide(10, 2))
print(divide(10, 0))
```

## 3. Handle More Than One Exception
Catch each expected exception separately so the response is clear.

```python
def divide_with_name(a, b):
    try:
        return a / b
    except ZeroDivisionError as error:
        print("ZeroDivisionError:", error)
    except TypeError as error:
        print("TypeError:", error)

print(divide_with_name(10, 2))
print(divide_with_name(10, 0))
print(divide_with_name(10, "two"))
```

## 4. Raise an Exception
Use `raise` when a function detects an invalid input that its caller should handle.

```python
def require_positive(n):
    if n <= 0:
        raise ValueError("n must be positive")
    return n

try:
    print(require_positive(-3))
except ValueError as error:
    print(error)
```

## 5. `try`, `except`, `else`, and `finally`
`else` runs when no exception occurs. `finally` runs whether or not an exception occurs.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as error:
        print("Error:", error)
    else:
        print("Result:", result)
    finally:
        print("Finished divide operation")

safe_divide(10, 2)
safe_divide(10, 0)
```

## Key Takeaways
- Exceptions represent runtime errors
- Catch specific exceptions rather than using a broad `except`
- Use `raise` to report invalid inputs to a caller
- `else` runs after a successful `try`; `finally` always runs
