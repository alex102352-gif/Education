# Python Sets



This notebook introduces sets and the four main set operations.

## 1. Create a Set
Sets store unique values. Repeated values appear only once.

```python
values = {1, 2, 3, 3, 3}

print(values)
```

## 2. Create Two Sets
We will compare a set of prime numbers with a set of odd numbers.

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}

print(primes)
print(odds)
```

## 3. Union
A union contains every unique value from both sets.

```python
print(primes.union(odds))
print(primes | odds)
```

## 4. Intersection
An intersection contains only values that appear in both sets.

```python
print(primes.intersection(odds))
print(primes & odds)
```

## 5. Difference
A difference removes from the first set any values that also occur in the second set. Order matters.

```python
print(odds.difference(primes))
print(odds - primes)

print(primes.difference(odds))
print(primes - odds)
```

## 6. Symmetric Difference
A symmetric difference contains values that appear in exactly one of the sets.

```python
print(primes.symmetric_difference(odds))
print(primes ^ odds)
```

## 7. Use Set Operators
Use `|`, `&`, `-`, and `^` for set operations. Do not use the logical keywords `and` and `or` for this purpose.

```python
print(primes & odds)
print(primes | odds)

# primes and odds checks truthiness; it does not calculate an intersection
```

## Key Takeaways
- Sets contain unique values
- Union, intersection, and symmetric difference are order-independent
- Difference is order-sensitive
- Use overloaded operators such as `|` and `&` for set operations
