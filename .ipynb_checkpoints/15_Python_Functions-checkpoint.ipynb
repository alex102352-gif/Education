# Python Functions



## 1. Define and Call a Function
Use `def` to define a reusable block of code. This function returns the first `n` Fibonacci numbers.

```python
def fibonacci(n):
    result = []
    a, b = 0, 1

    while len(result) < n:
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci(8))
```

## 2. Type Hints
Type hints make intended inputs and outputs clearer, but Python does not enforce them at runtime.

```python
def fibonacci_typed(n: int) -> list[int]:
    result = []
    a, b = 0, 1

    while len(result) < n:
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci_typed(8))
```

## 3. Return Multiple Values
Python returns multiple values as a tuple, which can be unpacked into variables.

```python
def fibonacci_with_max(n):
    values = fibonacci(n)
    return values, max(values)

values, largest = fibonacci_with_max(8)
print(values)
print(largest)
```

## 4. Default Parameters
A default value is used when the caller does not provide that argument.

```python
def fibonacci_start(n, a=0, b=1):
    result = []

    while len(result) < n:
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci_start(6))
print(fibonacci_start(6, 2, 3))
```

## 5. A Function Without a Return Value
Some functions only perform an action, such as modifying an existing list.

```python
def add_score(scores, score):
    scores.append(score)

scores = [85, 92]
add_score(scores, 98)
print(scores)
```

## 6. Flexible Arguments
`*args` collects unnamed arguments into a tuple. `**kwargs` collects named arguments into a dictionary.

```python
def show_arguments(*args, **kwargs):
    print(args)
    print(kwargs)

show_arguments(1, 2, 3, course="CPSC 5070", quarter="Fall")
```

## 7. Expand a Tuple and Dictionary
Use `*` and `**` when passing an existing tuple or dictionary into a function call.

```python
args = (1, 2, 3)
kwargs = {"course": "CPSC 5070", "quarter": "Fall"}

show_arguments(*args, **kwargs)
```

## 8. Pass a Function as an Argument
A sorting key function is applied to each item before sorting.

```python
groups = [[2, 3], [1, 2], [3, 4], [1, 100]]

print(sorted(groups))
print(sorted(groups, key=max))
```

## Key Takeaways
- Functions are defined with `def`
- `return` sends a value back to the caller
- Default parameters provide a fallback value
- `*args` and `**kwargs` support flexible argument lists
- Functions can be passed to other functions
