```python
x = [1, 2, 3]
print(x)
print(len(x))
```

```python
x.append(4)
print(x)

x.remove(2)
print(x)

x.insert(3, "a")
print(x)
```

```python
with_five = x + [5]
with_zero = [0] + x
print(with_five)
print(with_zero)
```

```python
values = [1, 2, 3, 2, 2]
print(values.count(2))
print(len(values))
```

```python
nested = [1, 2, [2, 3]]
print(nested.count(2))
print(nested.count([2, 3]))
```

```python
numbers = [3, 1.5, 2]
numbers.sort()
print(numbers)

nested_numbers = [[2, 3], [1, 3], [1, 2]]
nested_numbers.sort()
print(nested_numbers)
```

```python
x = [1, 2, 3, 4, 5, 6, 7]
print(x[0])
print(x[-1])
print(x[-2])
```

```python
print(x[:3])
print(x[3:])
print(x[-3:])
print(x[1:-3])
```

```python
print(x[::2])
print(x[::-1])
print(x[::-2])
```

```python
x[2:4] = ["a", "b", "c"]
print(x)

x[2:5] = ["d"]
print(x)
```

```python
numbers = [9, 4, 1, 6, 2]
numbers.append(7)
numbers.remove(4)
numbers.sort()
print(numbers[:3])
```
