## 1. Why are we using Anaconda and Jupyter?

In this course, we will use **Anaconda** and **Jupyter Notebook** as our main tools.

### Anaconda
Anaconda is a platform that comes with many of the libraries we need for data science already installed.

Some examples include:
- **NumPy** for numerical computing
- **Matplotlib** for plotting
- **Seaborn** for data visualization

That means you do not have to install each library one by one at the beginning of the course.

### Jupyter Notebook
Jupyter Notebook gives us an interactive place to:
- write Python code
- run code one piece at a time
- see results immediately
- mix code with explanation and notes

This makes it a great tool for learning Python and data science.


## 2. Jupyter runs in your browser

When you open Jupyter Notebook, it usually opens in your default web browser.

Even though it runs in a browser, it does **not** mean you need the internet for it to work.  
Jupyter can run locally on your computer using a local server.

That is why you may see something like:
- `localhost`
- a browser tab opening automatically
- a command prompt or terminal window running in the background

If the browser does not open automatically, you can copy the full URL from the command prompt, including the token.

## 4. First Python example: variables

A variable stores a value so we can use it later.

In Python, assignment uses the `=` sign.


```
x = 1
```

```
x
```

When you run the first cell, Python stores the value `1` in the variable `x`.

When you run the second cell, Jupyter displays the current value of `x`.


## 5. Running cells in Jupyter

Two useful shortcuts:

- **Shift + Enter** → runs the current cell and moves to the next cell
- **Alt + Enter** → runs the current cell and creates a new empty cell below

These shortcuts make Jupyter very convenient for step-by-step work.


## 6. Comments in Python

Comments are notes for humans.  
They help explain code, and Python ignores them during execution.

In Python, comments begin with the `#` symbol.


```
# This is a comment
x = 5  # This comment is at the end of a line
x
```

## 7. Python is easy to start with

Unlike some other programming languages, Python lets you begin writing code right away.

You do **not** need to:
- define a class first
- define a main function first
- add extra structure before testing simple ideas

This makes Python beginner-friendly and excellent for experimentation.


## 8. Semicolons in Python

In many languages, semicolons are required at the end of each statement.

In Python:
- semicolons are usually **not needed**
- they are optional at the end of a line
- they are mostly useful only when writing multiple statements on one line


```
a = 10
b = 20
print(a)
print(b)
```

```
a = 10; b = 20; print(a); print(b)
```

Both cells above work.

However, the first style is much clearer and is the preferred Python style.


## 9. Code blocks in Python: indentation matters

In languages like Java or C, code blocks are often shown with curly braces.

In Python, code blocks are shown using:
- a colon `:`
- indentation

This is one of Python’s most important syntax rules.


```
x = 7

if x > 5:
    print("x is greater than 5")
```

Notice:
- the `if` statement ends with a colon
- the line below it is indented

That indentation tells Python that the `print` statement belongs inside the `if` block.


## 10. Example: split numbers around a midpoint

The transcript mentions a Python example that separates numbers by a midpoint value.

Here is a simple version of that idea.


```
numbers = [1, 3, 5, 7, 9, 11]
midpoint = 6

lower = []
higher = []

for n in numbers:
    if n < midpoint:
        lower.append(n)
    else:
        higher.append(n)

print("Numbers below midpoint:", lower)
print("Numbers at or above midpoint:", higher)
```

This example shows several important Python ideas at once:
- lists
- variables
- a `for` loop
- an `if` statement
- indentation
- printing results
