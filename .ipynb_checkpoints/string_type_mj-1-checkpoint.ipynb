## 1. What is a String?

A string is a sequence of characters. In Python, strings are used to store text.

Some examples:
- a word like `"hello"`
- a sentence like `"Python is fun"`
- even a single character like `"A"`

Unlike some other languages, Python does not have a separate `char` type. A single character is just a string of length 1.


```
x = "hello"
y = 'hello'

print(x)
print(y)
print(type(x))
print(type(y))
print(x == y)
```

## 2. Single Quotes and Double Quotes

Python lets you create strings using either:
- single quotes: `'hello'`
- double quotes: `"hello"`

Most of the time, both forms mean the same thing. Use whichever makes the string easier to write and read.


```
a = "Python"
b = 'Python'

print(a)
print(b)
print(a == b)
```

## 3. Why Both Kinds of Quotes Are Useful

The real advantage appears when your string contains quotes inside it.

For example, suppose you want quotation marks inside the text itself.


```
sentence1 = 'He said, "Python is easy."'
sentence2 = "It's a nice day."

print(sentence1)
print(sentence2)
```

Using opposite quote types helps avoid confusion.

- If the text contains double quotes inside, wrap the whole string in single quotes.
- If the text contains a single quote inside, wrap the whole string in double quotes.


## 4. What Goes Wrong with Matching Quotes

If you start a string with double quotes, Python expects the next double quote to end it.  
That is why this kind of code causes problems:

`"He said, "Python is easy.""`

Python cannot tell which quotes belong to the string and which quotes are part of the text.


```
# Uncomment the next line to see the syntax problem
# bad_example = "He said, "Python is easy."" 
```

## 5. Escape Characters

Another way to include special characters inside a string is to use the backslash `\`.

The backslash is called an escape character. It tells Python to treat the next character specially.


```
s1 = 'It\'s a nice day.'
s2 = "He said, \"Python is easy.\""

print(s1)
print(s2)
```

This is useful when you want to keep the same outer quote style but still include that same quote inside the string.


## 6. Triple-Quoted Strings

Python also supports triple quotes for strings.

These are especially helpful for:
- multi-line text
- longer blocks of writing
- strings that contain both single and double quotes


```
message = """Welcome to Python.
This is a multi-line string.
It can span several lines easily."""

print(message)
```

Triple-quoted strings are convenient when your text naturally belongs on more than one line.


## 7. Length of a String

Use the built-in `len()` function to find how many characters are in a string.


```
x = "hello"

print(len(x))
print(len("Python"))
print(len("A"))
```

Remember:
- spaces count as characters
- punctuation counts as characters
- upper and lower case letters each count as one character


```
phrase = "Hello world!"
print(phrase)
print(len(phrase))
```

## 8. Strings Are Immutable

A very important idea: strings do not change in place.

When you use a string method like `.upper()` or `.capitalize()`, Python creates a new string instead of modifying the original one.


```
x = "hello"

print(x.upper())
print(x)
```

The original string is still `"hello"`.

That means `.upper()` returned a new string, but it did not rewrite `x`.


```
x = "hello"

y = x.upper()

print("x =", x)
print("y =", y)
```

## 9. Useful String Methods

Here are a few simple examples.


```
x = "hello"

print(x.upper())
print(x.capitalize())
print(x.title())
```

Try using Shift+Tab inside Jupyter after a method name like `x.upper(` to see documentation.

That is a good habit when you are learning.


## 10. Concatenation

Concatenation means joining strings together.

In Python, the `+` operator combines strings.


```
first = "hello"
second = " world"

result = first + second
print(result)
```

You can also build longer strings this way.


```
greeting = "Hello"
name = "Muneeba"

full = greeting + ", " + name + "!"
print(full)
```

## 11. Repetition

Python lets you repeat a string by multiplying it by an integer.


```
x = "ha"
print(x * 3)
print(3 * x)
```

This works because Python knows how to repeat a string a whole number of times.


```
word = "hello"
print(word * 4)
```

## 12. What Does Not Work

A string can be repeated by an integer, but not by a float.


```
# Uncomment to test the error
# "hello" * 3.5
```

If you really need a value like `3.5`, you would have to convert it first, but notice that converting to `int` truncates the decimal part.


```
x = "hello"
n = 3.5

print(int(n))
print(x * int(n))
```

## 13. Indexing

Strings are ordered sequences, so you can access individual characters by index.

Python uses zero-based indexing:
- index 0 is the first character
- index 1 is the second
- and so on


```
x = "hello"

print(x[0])
print(x[1])
print(x[4])
```

For the string `"hello"`:

- `x[0]` is `'h'`
- `x[1]` is `'e'`
- `x[4]` is `'o'`


## 14. Out-of-Range Indexing

If you ask for a character that does not exist, Python raises an error immediately.


```
x = "hello"

# Uncomment to see the error
# print(x[5])
```

This is actually helpful. Python catches the mistake early, which makes debugging easier.


## 15. A Few More Useful Examples


```
text = "python"

print("Original:", text)
print("Uppercase:", text.upper())
print("Capitalized:", text.capitalize())
print("Repeated twice:", text * 2)
print("Length:", len(text))
print("First letter:", text[0])
print("Last letter:", text[5])
```

Key ideas from this notebook:

- Python allows single quotes, double quotes, and triple quotes
- Use opposite quote styles or escape characters when needed
- `len()` gives the number of characters
- String methods return new strings
