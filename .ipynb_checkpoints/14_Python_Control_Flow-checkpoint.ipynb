# Python Control Flow


This notebook introduces conditions, `for` loops, `while` loops, `break`, and the loop `else` clause.

## 1. `if`, `elif`, and `else`
Use `elif` for an additional condition between `if` and `else`.

```python
x = 10

if x == 0:
    print("x is zero")
elif x > 0:
    print("x is positive")
else:
    print("x is negative")
```

## 2. A `for` Loop with `range()`
`range(6)` supplies the values from `0` through `5`.

```python
for i in range(6):
    print(i)
```

## 3. Loop Through Collections
Lists, tuples, dictionaries, and sets are iterable. Looping through a dictionary gives its keys.

```python
items = ["apple", "banana", "cherry"]
for item in items:
    print(item)

scores = {"Aden": 95, "Elyana": 98}
for name in scores:
    print(name, scores[name])

for number in {2, 3, 5, 7}:
    print(number)
```

## 4. A `while` Loop
A `while` loop continues as long as its condition is true.

```python
i = 0

while i < 10:
    print(i)
    i += 1
```

## 5. Stop a Loop with `break`
`break` immediately exits the current loop.

```python
for n in range(1, 100):
    if n % 10 == 0:
        break
    print(n)
```

## 6. A Loop `else` Clause
The `else` clause runs only when the loop finishes without a `break`.

```python
primes = []

for n in range(2, 31):
    for p in primes:
        if n % p == 0:
            break
    else:
        primes.append(n)

print(primes)
```

## Key Takeaways
- Use `if`, `elif`, and `else` to make decisions
- Use a `for` loop to move through an iterable
- Use a `while` loop when repetition depends on a condition
- `break` exits a loop, while loop `else` runs only when no break happens
