# Python Dictionaries



This notebook introduces dictionaries, key-value pairs, valid keys, views, and nested access.

## 1. Create and Access a Dictionary
A dictionary stores key-value pairs. Use a key, not an index, to retrieve a value.

```python
x = {"a": 1, "b": 2, "c": 3}

print(x)
print(x["a"])
print(x["b"])
```

## 2. Add and Update Values
Assigning a new key adds an entry. Assigning an existing key updates its value.

```python
x["d"] = 4
print(x)

x["a"] = 5
print(x)
```

## 3. Keys and Values Views
`keys()` and `values()` return views that reflect later changes to the dictionary.

```python
keys = x.keys()
values = x.values()
print(keys)
print(values)

x["e"] = 6
print(keys)
print(values)
```

## 4. Valid Dictionary Keys
Dictionary keys must be hashable. Strings, numbers, and tuples can be keys; lists cannot.

```python
locations = {(47.61, -122.33): "Seattle", 98101: "Downtown Seattle"}

print(locations[(47.61, -122.33)])
print(locations[98101])

# locations[[47.61, -122.33]] = "Seattle"  # Lists cannot be dictionary keys
```

## 5. Nested Structures
Use indexing for lists and tuples, then use keys to access dictionaries.

```python
data = [1, 2, ("x", "y", {"a": [1, 2, {"hello": "world"}]})]

print(data[2])
print(data[2][2])
print(data[2][2]["a"][2]["hello"])
```

## Key Takeaways
- Dictionaries store key-value pairs
- Access a dictionary with a key, not a position
- Assigning an existing key updates its value
- Keys must be hashable and cannot be mutable lists or dictionaries
