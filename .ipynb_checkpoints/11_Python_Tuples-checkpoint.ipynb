# Python Tuples


## 1. Create a Tuple
Tuples are similar to lists but are immutable. They commonly use parentheses.

```python
t = (1, 2, 3)

print(t)
print(type(t))
```

## 2. Read a Tuple
You can count, index, and slice a tuple just as you would a list.

```python
t = (1, 1, 2, 2, 3, 3)

print(t.count(2))
print(t[0])
print(t[:4])
```

## 3. Tuples Cannot Be Changed
Methods such as `append`, `insert`, and `remove` are not available for tuples.

```python
# t.append(4)
# t[0] = 10

print(t)
```

## 4. One-Item Tuples
A comma is required when a tuple has only one item.

```python
one_value = (1)
one_tuple = (1,)
without_parentheses = 1, 2, 3

print(type(one_value))
print(type(one_tuple))
print(type(without_parentheses))
```

## 5. Tuple Unpacking
Functions can return multiple values as a tuple, which can be assigned to multiple variables.

```python
x = 0.125
ratio = x.as_integer_ratio()
print(ratio)

numerator, denominator = ratio
print(numerator)
print(denominator)
```

## Key Takeaways
- Tuples are immutable
- Tuples support reading, indexing, slicing, and `count()`
- A one-item tuple needs a comma
- Tuples are useful for returning or unpacking multiple values
