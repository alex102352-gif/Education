# Python Lambda Functions


## 1. Create a Lambda Function
A lambda function is a small anonymous function. It can be assigned to a variable when needed.

```python
add = lambda x, y: x + y

print(add(1, 2))
print(add(1, 3))
```

## 2. Compare It with `def`
The following named function has the same behavior as the lambda above.

```python
def add_named(x, y):
    return x + y

print(add_named(1, 3))
```

## 3. Lambda as a Sorting Key
Use `key=` to tell `sort` or `sorted` how to compare each item.

```python
groups = [[2, 3], [1, 2], [3, 4], [1, 100]]

print(sorted(groups, key=min))
print(sorted(groups, key=max))
```

## 4. Sort Dictionaries by a Value
Each item in this list is a dictionary. The lambda returns the year of birth used for sorting.

```python
people = [
    {"first": "Grace", "last": "Hopper", "yob": 1906},
    {"first": "Alan", "last": "Turing", "yob": 1912},
    {"first": "Ada", "last": "Lovelace", "yob": 1815}
]

by_year = sorted(people, key=lambda item: item["yob"])
print(by_year)
```

## 5. Sort by Multiple Values
Return a tuple from the lambda to sort first by one value and then by another.

```python
people.append({"first": "Ada", "last": "Byron", "yob": 1815})

by_name = sorted(people, key=lambda item: (item["first"], item["last"]))
print(by_name)
```

## 6. Use a Lambda Once
Lambda functions are especially useful when a short function is needed only once.

```python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda n: n ** 2, numbers))
print(squared)
```

## Key Takeaways
- A lambda function is an anonymous function
- Lambda functions are useful for short, one-time operations
- `key=lambda ...` is common when sorting complex data
- A lambda can return a tuple for multiple sorting criteria
